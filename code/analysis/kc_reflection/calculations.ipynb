```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.constants import h
from prefixed import Float
```

```python
input_power = 310e-6
offset = 11.40e-9
```

```python
stage_1_output = 126.0e-9
stage_2_output = 126.1e-9
```

```python
reduction_stage_1 = (stage_1_output-offset)/(input_power-offset)
reduction_stage_2 = (stage_2_output-offset)/(input_power-offset)
```

```python
total_reduction = reduction_stage_1*reduction_stage_2
```

```python
center_freq = 384228.115e9
freq_upper = center_freq + 150.0e6
freq_lower = center_freq - 150.0e6
```

```python
mean_photon_number = 0.2
pulse_length = 0.372e-6
```

```python
def calculate_power(freq):
    return (mean_photon_number*h*freq)/pulse_length
```

```python
power_center = calculate_power(center_freq)/total_reduction
```

```python
power_upper = calculate_power(freq_upper)/total_reduction
power_lower = calculate_power(freq_lower)/total_reduction
```
